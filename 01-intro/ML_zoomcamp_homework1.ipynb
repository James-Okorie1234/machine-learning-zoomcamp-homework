#### Q7. Sumof weights

```python
# All the cars in Asia
# Had this variable already 
asia_cars.head()
```

```python
# Selecting vehicle weight and model number (first 7 values)
asia_cars[['vehicle_weight', 'model_year']].head(7)
```

```python
# Getting the underlying NumPy array
X = asia_cars[['vehicle_weight', 'model_year']].head(7).values
X
```

```python
XTX = X.T.dot(X) # Transpose of X multiplied by X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX) # Using Numpy linear algebra method to compute an inverse
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200]) 
y
```

```python
a = XTX_inv.dot(X.T) # multiplication of the inverse and the transpose of X
w = a.dot(y) # the output above multiplied by y
w
```

```python
# Sum of the elements in the results above
result_sum = w.sum()
result_sum
```
